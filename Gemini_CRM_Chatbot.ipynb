```python
import os
from google import genai
```

```python
# Configrations
API_KEY = "AIzaSyDJ1XjKeNtBaOni4LLuboIr4Ix9v_77_CM"
MODEL_NAME = "gemini-2.5-flash"  # latest flash model :contentReference[oaicite:1]{index=1}
HISTORY_FILE = "chat_history.txt"
```

```python
# Adding api key
client = genai.Client(api_key=API_KEY)
```

```python
# Load previous history to maintain contexts
history_text = ""
if os.path.exists(HISTORY_FILE):
    with open(HISTORY_FILE, "r", encoding="utf-8") as f:
        history_text = f.read().strip()
```

```python
# CRM system instruction text for context it will give the output according to CRM
crm_instruction = (
    "You are a CRM assistant helping the user with customer queries, "
    "sales followups, account management, and CRM‑related explanations. "
    "Answer concisely and professionally."
)
```

```python
# Create chat session

# If there’s existing history, we put it in the first message
initial_history = []
if history_text:
    initial_history.append({
        "role": "user",
        "parts": [{"text": history_text}]
    })
```

```python
#creating Chat configs
chat = client.chats.create(
    model=MODEL_NAME,
    history=initial_history
)
```

```python
# main chatbot
print("CRM Chatbot (type 'exit' to quit)\n")

while True:
    user_input = input("\nYou: ")
    if user_input.lower() == "exit":
        print("Goodbye!")
        break

    # Prepend system instruction to each prompt
    prompt = crm_instruction + "\nUser: " + user_input

    # Send the message
    response = chat.send_message(prompt)

    reply_text = response.text
    print("\n\nCRM Bot:", reply_text)

    # Save conversation locally
    with open(HISTORY_FILE, "a", encoding="utf-8") as f:
        f.write(f"You: {user_input}\n")
        f.write(f"CRM Bot: {reply_text}\n")
```
